```python
from ultralytics import YOLO
```

```python

model = YOLO("yolov8n.pt")
model.train(data="db/sudoku/data.yaml", epochs=10,imgsz=300)  # train the model
```

```python
results = model("media/ex1.jpg", show=False, save=False)
result = results[0]  # Get the first result for a single image

# Loop through each detected object
for box in result.boxes:
    # 1. Get the accuracy / confidence score (as a float)
    accuracy = float(box.conf[0])
    
    # 2. Get the bounding box coordinates [xmin, ymin, xmax, ymax]
    # .xyxy returns coordinates in pixels
    bbox = box.xyxy[0].tolist() 
    
    print(f"Accuracy: {accuracy:.4f} | Bounding Box: {bbox}")
```
